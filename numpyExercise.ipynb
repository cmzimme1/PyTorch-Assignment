Create:

````python
x = np.array([[12, 5, 18, 3],
              [25, 7, 14, 9],
              [2, 30, 11, 20]])
print("1. Values greater than 15:", x[x > 15])
print("2. Values at most 10:", x[x <= 10])
original_count = np.count_nonzero(x > 15)

x[x > 20] = 0
print("3. After replacing values greater than 20:\n", x)
print("4. Count greater than 15 after replacement:", np.count_nonzero(x > 15))
print("Count greater than 15 in the original array:", original_count)
````

## Part 5 — Matrix Multiplication
Create:
```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[7, 8],
              [9, 10],
              [11, 12]])
```
1.	Compute A @ B.
2.	Compute the same result using np.dot(A, B).
3.	Verify that both results are identical.
4.	Print the shape of A, B, and the result.

The notebook demonstrates the equivalence of .dot(), np.dot(), and @ for matrix multiplication.

````python
A = np.array([[1, 2, 3], [4, 5, 6]])
B = np.array([[7, 8], [9, 10], [11, 12]])
product = A @ B
product_dot = np.dot(A, B)
print("1. A @ B:\n", product)
print("2. np.dot(A, B):\n", product_dot)
print("3. Results identical:", np.array_equal(product, product_dot))
print("A.dot(B) also matches:", np.array_equal(product, A.dot(B)))
print("4. Shapes: A =", A.shape, "B =", B.shape, "result =", product.shape)
````

## Part 6 — Transpose and Axis Operations
Using:
```python
X = np.array([[2, 4, 6],
              [8, 10, 12]])
```
Calculate:
1.	X.T
2.	np.transpose(X)
3.	np.sum(X)
4.	np.sum(X, axis=0)
5.	np.sum(X, axis=1)
6.	np.mean(X)
7.	np.mean(X, axis=0)
8.	np.mean(X, axis=1)
9.	np.max(X, axis=0)
10.	np.min(X, axis=1)
11.	np.std(X)

For each axis operation, write one sentence explaining what the axis represents.

Axis explanations:

- `np.sum(X, axis=0)` adds down the rows, giving one sum for each column.
- `np.sum(X, axis=1)` adds across the columns, giving one sum for each row.
- `np.mean(X, axis=0)` averages down the rows, giving one mean for each column.
- `np.mean(X, axis=1)` averages across the columns, giving one mean for each row.
- `np.max(X, axis=0)` finds the largest entry down the rows in each column.
- `np.min(X, axis=1)` finds the smallest entry across the columns in each row.
- With no axis specified, sum, mean, and standard deviation use all elements.


````python
X = np.array([[2, 4, 6], [8, 10, 12]])
print("1. X.T:\n", X.T)
print("2. np.transpose(X):\n", np.transpose(X))
print("3. Total sum:", np.sum(X))
print("4. Column sums:", np.sum(X, axis=0))
print("5. Row sums:", np.sum(X, axis=1))
print("6. Overall mean:", np.mean(X))
print("7. Column means:", np.mean(X, axis=0))
print("8. Row means:", np.mean(X, axis=1))
print("9. Column maxima:", np.max(X, axis=0))
print("10. Row minima:", np.min(X, axis=1))
print("11. Overall standard deviation:", np.std(X))
````

## Part 7 — Mini Application: Student Grades

Create a NumPy array containing the grades of 5 students in 4 assignments:

* Student 1: 85, 90, 78, 92
* Student 2: 70, 88, 91, 76
* Student 3: 95, 93, 89, 97
* Student 4: 60, 72, 68, 75
* Student 5: 88, 84, 90, 86

Using NumPy only:
1.	Calculate each student’s average.
2.	Find the highest grade on each assignment.
3.	Find the average grade for each assignment.
4.	Find the student with the highest overall average.
5.	Find all grades below 70.
6.	Replace grades below 70 with 70.
7.	Calculate the overall class average.

````python
grades = np.array([[85, 90, 78, 92],
                   [70, 88, 91, 76],
                   [95, 93, 89, 97],
                   [60, 72, 68, 75],
                   [88, 84, 90, 86]])
student_averages = np.mean(grades, axis=1)
print("1. Student averages:", student_averages)
print("2. Highest grade per assignment:", np.max(grades, axis=0))
print("3. Average per assignment:", np.mean(grades, axis=0))
best_student = np.argmax(student_averages) + 1
print("4. Student with highest average:", best_student)
print("Highest average:", np.max(student_averages))
print("5. Grades below 70:", grades[grades < 70])
original_class_average = np.mean(grades)
# Apply the requested minimum grade before calculating step 7.
grades[grades < 70] = 70
print("6. Grades after replacement:\n", grades)
print("7. Overall class average after replacement:", np.mean(grades))
print("Original class average:", original_class_average)
````

## Challenge Problem

Create a random 5 × 5 matrix containing integers from 1 to 100.

Using NumPy only:
1.	Find the largest value.
2.	Find the smallest value.
3.	Calculate the mean.
4.	Calculate the standard deviation.
5.	Find all values greater than the mean.
6.	Replace all values below the mean with 0.
7.	Calculate the sum of each row and each column.

````python
challenge_rng = np.random.default_rng(42)
matrix = challenge_rng.integers(1, 101, size=(5, 5))
print("Original random matrix:\n", matrix)
mean = np.mean(matrix)
print("1. Largest value:", np.max(matrix))
print("2. Smallest value:", np.min(matrix))
print("3. Mean:", mean)
print("4. Standard deviation:", np.std(matrix))
print("5. Values greater than the original mean:", matrix[matrix > mean])

matrix[matrix < mean] = 0
print("6. Values below the mean replaced with zero:\n", matrix)
print("7. Row sums after replacement:", np.sum(matrix, axis=1))
print("Column sums after replacement:", np.sum(matrix, axis=0))
````
