Explain how PyTorch tensor indexing is similar to NumPy array indexing.

Question

Explain the difference between:
```python
A * B
and:
A @ B
```

`A * B` multiplies matching entries independently. `A @ B` performs matrix multiplication: each output entry is the sum of products of a row of A and a column of B. Here `A * B` is `[[5, 12], [21, 32]]`, while `A @ B` is `[[19, 22], [43, 50]]`.

Questions

Explain:

* What does dim=0 mean?
* What does dim=1 mean?
* Why does reducing along a dimension change the shape of the tensor?

- `dim=0` reduces down the rows, leaving one result per column.
- `dim=1` reduces across the columns, leaving one result per row.
- A reduction combines all entries along the selected dimension. By default, that dimension is removed: summing a `(2, 3)` tensor over dim 0 gives shape `(3,)`, and over dim 1 gives shape `(2,)`. `keepdim=True` keeps the reduced dimension with length 1.

Part 9 — Automatic Differentiation

PyTorch provides automatic differentiation using autograd.

````python
x = torch.tensor([[1., 2.], [3., 4.]], requires_grad=True)
y = 3 * x ** 2 + 2
print("y:\n", y)
````

````python
z = torch.sum(y)
z.backward()
print("y:\n", y)
print("z:", z)
print("x.grad:\n", x.grad)
print("Gradient matches 6x:", torch.allclose(x.grad, 6 * x.detach()))
````

Questions
* What does requires_grad=True mean?
* What does backward() do?
* What is stored in x.grad?
* Why is automatic differentiation important for neural networks?

- `requires_grad=True` tells PyTorch to track operations involving this tensor so it can calculate derivatives.
- `backward()` applies the chain rule through the computation graph and accumulates gradients in leaf tensors. Here z is scalar, so no explicit starting gradient is needed.
- `x.grad` stores the derivative of z with respect to each entry of x. Since z is the sum of `3x² + 2`, its gradient is `6x`, giving `[[6, 12], [18, 24]]`.
- Neural networks need gradients of the loss with respect to their parameters. Automatic differentiation calculates those gradients so an optimizer can update the parameters without manually deriving every derivative.
